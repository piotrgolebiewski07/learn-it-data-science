```python
# Zadanie 13
import numpy as np

# Tworzymy sdzereg czasowy
dates = pd.date_range('2026-10-01', periods=20, freq='D')
values = [120, 140, np.nan, np.nan, np.nan, 130, 150, 160, 170, 200, 190, np.nan, np.nan, 180, 170, np.nan, np.nan, 230, 250, np.nan]

df_time = pd.DataFrame({'data': dates, 'wartość': values})
df_time.set_index('data', inplace=True)

# Forward fill i backward fill
df_ffill = df_time.copy()
df_ffill['wartość'] = df_ffill['wartość'].ffill()

df_bfill = df_time.copy()
df_bfill['wartość'] = df_bfill['wartość'].bfill()
df_bfill

# Interpolacja liniowa
df_linear = df_time.copy()
df_linear['wartość'] = df_linear['wartość'].interpolate(method='linear')

# Zestwienie w jednej tabeli
compare = pd.DataFrame({
    'oryginał': df_time['wartość'],
    'ffill': df_ffill['wartość'],
    'bfill': df_bfill['wartość'],
    'linear': df_linear['wartość']
})

compare.round(2)

# Wykres tekstowy
for i in ['oryginał', 'ffill', 'bfill', 'linear']:
    print(f"{i}")
    for date, val in compare[i].items():
        if pd.isnull(val):
            bar = '(Nan)'
        else:
            bar = '█' * int(val // 2)
        print(f"   {date.date()} |{bar}")
        
    
```

```python
# Zadanie 15
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'

batch_size = 50

#zmienne akumulacyjne
total_passengers = 0
sum = 0
count = 0
survived_sum = 0
batch_memory = 0 

print("Wczytywanie w batchach:")
for i, chunk in enumerate(pd.read_csv(url, chunksize=batch_size)):
    print(f"Batch {i+1}: {len(chunk)} wierszy")
    total_passengers += len(chunk['Age'])
    sum += chunk['Age'].sum()
    count += chunk['Age'].count()
    survived_sum += chunk['Survived'].sum()
    if batch_memory < chunk.memory_usage(deep=True).sum():
        batch_memory = chunk.memory_usage(deep=True).sum()
    
print(f"Sum: {sum}") # liczba pasażerów ze znanym wiekiem
print(f"Count: {count}")
print(f"Survived: {survived_sum}")

print(f"\nTotal passengers: {total_passengers}")
print(f"avg_age: {round(sum / count, 2)}")
print(f"survival_rate: {round(survived_sum / total_passengers * 100, 2)}%")
print(f"batch memory: {round(batch_memory/1024, 2)}KB ")

print(f"\nStatystki z ładowaniem pełnego df")
df = pd.read_csv(url)
print(f"total_passengers: {len(df['Age'])}")
print(f"avg_age: {round(df['Age'].mean(), 2)}")
print(f"survival_rate: {round(df['Survived'].sum() / len(df['Age']) * 100, 2)}%")
print(f"batch memory: {round(df.memory_usage(deep=True).sum()/1024, 2)}KB ")
```

```python
# Zadanie 18
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic = pd.read_csv(url)


def clean_titanic(df):
    df_titanic = df.copy()
    df_titanic = df_titanic.dropna(thresh = (len(df_titanic) * 0.4), axis=1)

    df_titanic_age_median = df_titanic['Age'].median()
    df_titanic['Age'] = df_titanic['Age'].fillna(df_titanic_age_median)
    df_titanic_embarked_dominanta = df_titanic['Embarked'].mode()[0]
    df_titanic['Embarked'] = df_titanic['Embarked'].fillna(df_titanic_embarked_dominanta)
        
    df_titanic['PassengerId'] = df_titanic['PassengerId'].astype('int16')
    df_titanic['Survived'] = df_titanic['Survived'].astype('int8')
    df_titanic['Pclass'] = df_titanic['Pclass'].astype('int8')
    df_titanic['SibSp'] = df_titanic['SibSp'].astype('int8')
    df_titanic['Parch'] = df_titanic['Parch'].astype('int8')
    
    df_titanic['Sex'] = df_titanic['Sex'].astype('category')
    df_titanic['Embarked'] = df_titanic['Embarked'].astype('category')
    
    df_titanic['FamilySize'] = df_titanic['SibSp'] + df_titanic['Parch'] + 1
    return df_titanic

print(clean_titanic(titanic).info())
```

```python
# Zadanie 19
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic = pd.read_csv(url)

kwantyle = titanic['Fare'].quantile([0.25, 0.75])
Q1 = kwantyle.iloc[0]
Q3 = kwantyle.iloc[1]

# rozstęp międzykwartylowy
IQR = Q3 - Q1

# Dolna granica
dolna_granica = Q1 - 1.5 * IQR

# Górna granica
gorna_granica = Q3 + 1.5 * IQR

print(f"\nQ1 = {Q1}")
print(f"Q3 = {Q3}")
print(f"IQR = {IQR}")
print(f"\ndolna granica = {dolna_granica}")
print(f"górna granica = {gorna_granica}")

# Liczba outlierów
outliery = len(titanic[titanic['Fare'] > gorna_granica])
print(f"\nWszystkich outlierów jest: {outliery}")
print(f"Zakres wartości od {dolna_granica} do {gorna_granica}")

# liczba outlierów
outliery = titanic[titanic['Fare'] > gorna_granica]
print(f"\nLiczba outlierów: {len(outliery)}\n")

# paseżerowie bez outlierów
bez_outlierow = titanic[titanic['Fare'] <= gorna_granica]

# capping - przycinanie wartości do górnej granicy
titanic_capping = titanic.copy()
titanic_capping['Fare'] = np.clip(titanic['Fare'], dolna_granica, gorna_granica)

# transformacja logarytmiczna
titanic_log = titanic.copy()
titanic_log['Fare'] = np.log1p(titanic['Fare'])

# describe
print(f"\nDescribe: bez outlierów")
display(bez_outlierow.describe())

print(f"\nDescribe: capping")
display(titanic_capping.describe())

print(f"\nDescribe: transformacja logarytmiczna")
display(titanic_log.describe())

# Wnioski:
# - usunięcie outlierów zmniejsza wpływ wysokich cen biletów, ale przez to tracimy część danych
# - capping ogranicza maksymalne wartości do wyznaczonej granicy, nie usuwając przy tym żadnych pasażerów.
# - transformacja logarytmiczna zmniejsza różnice między małymi i dużymi wartościami i ta transormacja może dać najbardziej  symetryczny rozkład
```

```python
# Zadanie 20
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'

# wariant 1 - pełne wczytanie
titanic_1 = pd.read_csv(url)
memory_1 = round(titanic_1.memory_usage(deep=True).sum() / 1024, 2)

# wariant 2 - wczytanie wybranych kolumn
titanic_2 = pd.read_csv(url, usecols=['Name', 'Age', 'Sex', 'Pclass', 'Survived'])
memory_2 = round(titanic_2.memory_usage(deep=True).sum() / 1024, 2)

# wariant 3 - batch
batch_size = 100
memory_3 = 0

for chunk in pd.read_csv(url, chunksize=batch_size):
    if memory_3 < chunk.memory_usage(deep=True).sum() / 1024:
        memory_3 = round(chunk.memory_usage(deep=True).sum() / 1024, 2)                         

# wariant 4 - optymalizacja
titanic_4 = pd.read_csv(url)

titanic_4['Pclass'] = titanic_4['Pclass'].astype('int8')
titanic_4['Survived'] = titanic_4['Survived'].astype('int8')
titanic_4['SibSp'] = titanic_4['SibSp'].astype('int8')
titanic_4['Parch'] = titanic_4['Parch'].astype('int8')
titanic_4['Sex'] = titanic_4['Sex'].astype('category')
titanic_4['Embarked'] = titanic_4['Embarked'].astype('category')
memory_4 = round(titanic_4.memory_usage(deep=True).sum() / 1024, 2)

# zestawienie
results=pd.DataFrame(columns=['Metoda', 'Pamięć (KB)', 'Oszczędność (%)'])
results['Metoda'] = ['pełne wczytanie','wczytanie wybranych kolumn','batch','optymalizacja']
results['Pamięć (KB)'] = [memory_1, memory_2, memory_3, memory_4]

# oszczędność
for i in range(4):
    if i == 0:
        results.iloc[i, 2] = 0
    else:
        results.iloc[i, 2] = round(((results.iloc[0, 1] - results.iloc[i, 1]) * 100 / results.iloc[0, 1]), 2)
        
display(results)


# rysunek
jednostka = int(results.iloc[0,1] // 10)
print(f"\nZestawienie graficzne")
for i in range(4):
    bar = '█' * (int(results.iloc[i,1] // jednostka))
    print(f"- {results.iloc[i,0]:<28} {bar}")

# oszczędność energii
wartosc = 0.0003
for i in range(4):
    wynik = ((results.iloc[0,1] - results.iloc[i,1]) / 1024) * wartosc
    print(f"Oszczędnosć energii wariantu {i+1} względem pierwszego wariantu wynosi: {wynik:.8f}kWh/h")
```
